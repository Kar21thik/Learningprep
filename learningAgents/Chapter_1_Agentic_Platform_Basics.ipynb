## 1️⃣ What is an Agentic Platform?

### Real-World Analogy: A Customer Support Team

```
Traditional Support:
Customer → Single Agent → Fixed Questions → Limited Help

Agentic Platform:
Customer → Smart Agent → Can Think → Can Use Tools → Better Help
                ↓
         - Search database
         - Call APIs
         - Talk to other agents
         - Make decisions
```

### Key Difference:

| Traditional System | Agentic Platform |
|---|---|
| "Here's the answer" | "Let me think about this..." |
| Pre-written responses | Dynamic decisions |
| Can't use tools | Can call external services |
| No memory | Remembers conversation |
| Single purpose | Multi-purpose |

## 2️⃣ Core Concepts

### Concept 1: What is an AGENT?

Think of it like: **A virtual employee with a job description**

```python
# Agent definition
agent = {
    "id": "learning_coach_agent",
    "name": "Learning Coach",
    "description": "Helps users learn new topics",
    
    # Job description (how to behave)
    "instructions": "You are a helpful learning coach...",
    
    # Tools they can use
    "tools": [
        "search_web",
        "fetch_documentation",
        "save_notes"
    ],
    
    # Brain settings
    "llm_model": "claude-sonnet-4.6",
    "temperature": 0.7,
    "max_tokens": 2048
}
```

**Agent has:**
- ✓ Identity (name, description)
- ✓ Instructions (personality, rules)
- ✓ Tools (capabilities)
- ✓ Memory (conversation history)
- ✓ Settings (temperature, tokens)

### Concept 2: What are TOOLS?

Think of it like: **A mechanic's toolbox**

```
Agent wants to search the web
  ↓
Agent calls: search_web("Python lists tutorial")
  ↓
Tool executes (via MCP protocol)
  ↓
Returns: [search results]
  ↓
Agent uses result in response
```

**Common Tools:**
- 🔍 Search tools
- 📊 Database queries
- 📧 Email/messaging
- 💾 File operations
- 🌐 API calls

### Concept 3: What is a SESSION?

Think of it like: **A WhatsApp conversation thread**

```
Session ID: session-001

Message 1: User → "Explain Python lists"
Message 2: Agent → "Python lists are collections..."
Message 3: User → "Give an example"
Message 4: Agent → "Example: [1, 2, 3]"
Message 5: User → "How do I add items?"
Message 6: Agent → "Use append()..."

← Agent remembers ALL 6 messages!
```

**Benefits:**
- ✓ Continuity (agent remembers context)
- ✓ Personalization (knows user's journey)
- ✓ Multi-turn conversations
- ✓ Persistent history

## 3️⃣ How the Agentic Platform Works - Step by Step

### The Complete Request Flow

```
┌─────────────────────────────────────────┐
│ STEP 1: User Sends Request              │
│ "learning_coach_agent, explain Python" │
└─────────┬───────────────────────────────┘
          │
          ↓
┌─────────────────────────────────────────┐
│ STEP 2: API Receives Request            │
│ /agent/invoke/learning_coach_agent      │
└─────────┬───────────────────────────────┘
          │
          ↓
┌─────────────────────────────────────────┐
│ STEP 3: Load Agent from Database        │
│ Query DynamoDB for agent config         │
└─────────┬───────────────────────────────┘
          │
          ↓
┌─────────────────────────────────────────┐
│ STEP 4: Load Conversation History       │
│ Previous messages from this session     │
└─────────┬───────────────────────────────┘
          │
          ↓
┌─────────────────────────────────────────┐
│ STEP 5: Create Tools                    │
│ Connect to MCP servers                  │
└─────────┬───────────────────────────────┘
          │
          ↓
┌─────────────────────────────────────────┐
│ STEP 6: Send to LLM                     │
│ Send: instructions + history + tools    │
└─────────┬───────────────────────────────┘
          │
          ↓
┌──────────────────────────┐
│ STEP 7: Claude Thinks    │
│ Decision: Use tool? OR   │
│ Respond directly?        │
└──────────┬───────────────┘
           │
    ┌──────┴──────┐
    │             │
    ↓             ↓
[Use Tool]  [Answer Direct]
    │             │
    ↓             │
Call MCP Tool    │
Get Result       │
    │             │
    └──────┬──────┘
           ↓
┌──────────────────────────────┐
│ STEP 8: Generate Response    │
│ Claude creates final answer  │
└──────────┬───────────────────┘
           ↓
┌──────────────────────────────┐
│ STEP 9: Save to History      │
│ Store user + agent messages  │
└──────────┬───────────────────┘
           ↓
┌──────────────────────────────┐
│ STEP 10: Return Response     │
│ Send back to client          │
└──────────────────────────────┘
```

## 5️⃣ Local Setup Status

````python
import subprocess
import json

print("🔍 Checking Local Setup...\n")

# Check Docker containers
result = subprocess.run(
    ['docker', 'ps', '--format', 'table {{.Names}}\t{{.Status}}'],
    capture_output=True,
    text=True
)

print("✓ Running Containers:")
print(result.stdout)

# Check API
try:
    result = subprocess.run(
        ['curl', '-s', 'http://localhost:8080/agent/oob/agents'],
        capture_output=True,
        text=True,
        timeout=5
    )
    data = json.loads(result.stdout)
    print(f"\n✓ API Working: Found {len(data.get('agents', []))} agents")
    for agent in data.get('agents', []):
        print(f"  - {agent['agent_id']}: {agent['name']}")
except Exception as e:
    print(f"✗ API Error: {e}")
````

## 6️⃣ Key Concepts Summary

### Quick Reference

| Concept | What | Purpose |
|---------|------|----------|
| **Agent** | Virtual employee | Handle user requests |
| **Tool** | Capability | Execute tasks (search, DB, API) |
| **Session** | Conversation thread | Remember context |
| **LLM** | Claude model | Think and decide |
| **MCP** | Tool protocol | Call external services |
| **DynamoDB** | Database | Store agents & history |
| **Redis** | Cache | Fast access to sessions |

### Quick Commands

```bash
# Start services
docker run -d --name redis-local -p 6379:6379 redis:7
docker run -d --name dynamodb-local -p 21003:8000 amazon/dynamodb-local

# Run platform
cd /agentic-platform && uv run python main.py

# Test API
curl http://localhost:8080/agent/oob/agents
```
